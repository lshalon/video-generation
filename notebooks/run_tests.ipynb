# Run Tests

Run the pipeline test suite from this notebook. All external API calls (Claude, Fal) are mocked so these run offline without API keys.

```python
import subprocess
from IPython.display import Markdown
```

## Run all tests

```python
result = subprocess.run(
    ["python", "-m", "pytest", "-v", "--tb=short"],
    capture_output=True,
    text=True,
    cwd="..",
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```

## Run by category

Pick a specific test module to run.

```python
# Choose which test file to run:
# - "tests/test_config.py"                    Config, CLI parsing, dataclasses
# - "tests/test_analyze_reference.py"          Frame extraction + Claude analysis
# - "tests/test_write_script.py"               Product images + script generation
# - "tests/test_generate_starting_frame.py"    Two-stage image generation
# - "tests/test_generate_video.py"             Kling / Seedance video generation
# - "tests/test_pipeline.py"                   Pipeline orchestration

TEST_FILE = "tests/test_pipeline.py"  # <-- change this

result = subprocess.run(
    ["python", "-m", "pytest", "-v", "--tb=short", TEST_FILE],
    capture_output=True,
    text=True,
    cwd="..",
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```

## Run with coverage

```python
result = subprocess.run(
    ["python", "-m", "pytest", "--cov=src/video_generation", "--cov-report=term-missing", "-v", "--tb=short"],
    capture_output=True,
    text=True,
    cwd="..",
)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
```

## Lint check

```python
result = subprocess.run(
    ["python", "-m", "ruff", "check", "src/", "tests/"],
    capture_output=True,
    text=True,
    cwd="..",
)
print(result.stdout or "All checks passed!")
if result.returncode != 0:
    print(result.stderr)
```
